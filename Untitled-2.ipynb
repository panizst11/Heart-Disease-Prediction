```python
import pandas as pd
import numpy as np
from sklearn import preprocessing
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

disease_df = pd.read_csv("D:/project./2/framingham (1).csv")
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 10))

corr = disease_df.corr()

sns.heatmap(corr,annot=True,fmt=".2f",cmap="coolwarm",center=0)

plt.title("Correlation Matrix")
plt.show()
corr_with_target = disease_df.corr()['TenYearCHD'].sort_values(ascending=False)

print(corr_with_target)
disease_df.drop(columns=['education'], inplace = True, axis = 1)
disease_df.rename(columns ={'male':'Sex_male'}, inplace = True)
```

```python
disease_df.dropna(axis = 0, inplace = True)
disease_df
```

```python
print(disease_df.TenYearCHD.value_counts())
```

```python

X = np.asarray(disease_df[['age', 'Sex_male', 'cigsPerDay','totChol', 'sysBP', 'glucose']])
y = np.asarray(disease_df['TenYearCHD'])
from sklearn.model_selection import train_test_split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=42,stratify=y)
X_train, X_val, y_train, y_val = train_test_split(X_train,y_train,test_size=0.2,random_state=42,stratify=y_train)
print ('val set:', X_val.shape,  y_val.shape)
print ('Train set:', X_train.shape,  y_train.shape)
print ('Test set:', X_test.shape,  y_test.shape)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(x='TenYearCHD', data=disease_df,
             palette="BuGn_r")
plt.show()
```

```python
laste = disease_df['TenYearCHD'].plot()
plt.show(laste)
```

```python
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression(class_weight='balanced')
logreg.fit(X_train, y_train)
y_pred = logreg.predict(X_test)
```

```python
from sklearn.metrics import f1_score
y_prob = logreg.predict_proba(X_val)[:, 1]
best_threshold = 0
best_f1 = 0

for threshold in [i/100 for i in range(1, 100)]:
    
    y_VALpred = (y_prob >= threshold).astype(int)
    
    f1 = f1_score(y_val, y_VALpred)
    
    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold

print("Best threshold:", best_threshold)
print("Best F1:", best_f1)
```

```python
from sklearn.metrics import accuracy_score
print('Accuracy of the model is =', 
      accuracy_score(y_val, y_VALpred))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print('The details for confusion matrix is =')
print (classification_report(y_val, y_VALpred))

cm = confusion_matrix(y_val, y_VALpred)
conf_matrix = pd.DataFrame(data = cm,columns = ['Predicted:0', 'Predicted:1'], index =['Actual:0', 'Actual:1'])

plt.figure(figsize = (8, 5))
sns.heatmap(conf_matrix, annot = True, fmt = 'd', cmap = "Greens")

plt.show()
```

```python
y_test_prob = logreg.predict_proba(X_test)[:, 1]

y_test_pred = (y_test_prob >= best_threshold).astype(int)
```

```python
from sklearn.metrics import f1_score

test_f1 = f1_score(y_test, y_test_pred)

print("Test F1:", test_f1)
```

```python
print(classification_report(y_test, y_test_pred))
```
